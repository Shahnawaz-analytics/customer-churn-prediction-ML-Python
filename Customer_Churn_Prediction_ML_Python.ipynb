```python
# Dropping customerID column as it does not contribute to prediction
df.drop('customerID', axis=1, inplace=True)
```

```python
df.head()
```

```python
# Encoding categorical variables correctly
# Using pd.get_dummies() for nominal columns (no ordinal relationship)
# LabelEncoder is only appropriate for the target variable (Churn), not features

df = pd.get_dummies(df, drop_first=True)

print("Shape after encoding:", df.shape)
df.head()
```

```python
x=df.drop('Churn_Yes', axis=1)
y=df['Churn_Yes'].astype(int)
```

```python
# Splitting the dataset into Training and Testing sets

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Displaying training and testing dataset dimension

print("x_train Shape:", x_train.shape)
print("x_test Shape:", x_test.shape)
print("y_train Shape:", y_train.shape)
print("y_test Shape:", y_test.shape)
```

```python
# Standardizing numerical features for better model performance

from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()

x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
x_test_scaled
```

```python
# Train Logistic Regression Model with class_weight='balanced'
# This handles class imbalance by penalizing misclassification of minority class (churners)

from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42)

lr.fit(x_train_scaled, y_train)
```

```python
# Generating prediction using Logistic Regression model
lr_pred=lr.predict(x_test_scaled)
```

```python
# Evaluating Logistic Regression model accuracy

from sklearn.metrics import accuracy_score

lr_accuracy=accuracy_score(y_test, lr_pred)

print("Logistic Regression Accuracy:", lr_accuracy)
```

```python
# Generating detailed classification metrics
from sklearn.metrics import classification_report

print(classification_report(y_test, lr_pred))
```

```python
# Creating confusion matrix for model evaluation
from sklearn.metrics import confusion_matrix

cm=confusion_matrix(y_test, lr_pred)

cm
```

```python
# visualizing cofusion matrix using heatmap
sns.heatmap(cm, annot=True, fmt='d', cmap='viridis')
plt.title('Logistic Regression Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
# Training Random Forest classifier with class_weight='balanced'

from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(random_state=42, class_weight='balanced')

rf_model.fit(x_train, y_train)
```

```python
# Generating predictions using Random Forest model
rf_pred=rf_model.predict(x_test)
```

```python
# Evaluating Random Forest model accuracy

rf_accuracy=accuracy_score(y_test, rf_pred)

print("Random Forest Accuracy:", rf_accuracy)
```

```python
# comparing model performances
print("Logistic Regression Accuracy:", lr_accuracy)
print("Random Forest Accuracy:", rf_accuracy)
```

```python
# Generating classification report for Random Forest model
print(classification_report(y_test, rf_pred))
```

```python
# Creating confusion matrix for Random Forest
from sklearn.metrics import confusion_matrix

rf_cm = confusion_matrix(y_test, rf_pred)

sns.heatmap(rf_cm, annot=True, fmt='d', cmap='viridis')
plt.title('Random Forest Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()


```

```python
# Indentifying the most important features influencing churn
feature_importance=pd.DataFrame({'Feature':x.columns, 'Importance':rf_model.feature_importances_})

feature_importance=feature_importance.sort_values(by='Importance', ascending=False)

feature_importance.head(10)
```

```python
# Visualizing top 10 important features
top_features=feature_importance.head(10)

plt.figure(figsize=(8,5))
sns.barplot(x='Importance', y='Feature', data=top_features)
plt.title('Top 10 Important Features')
plt.show()
```

```python
# Hyperparameter tuning using GridSearchCV (proper method vs guessing values)
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [5, 10, 15],
    'min_samples_split': [2, 5, 10]
}

grid_search = GridSearchCV(
    RandomForestClassifier(random_state=42, class_weight='balanced'),
    param_grid,
    cv=5,                    # 5-fold cross-validation
    scoring='recall',        # Optimizing for recall — catching churners matters more
    n_jobs=-1,
    verbose=1
)

grid_search.fit(x_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best CV Recall Score:", round(grid_search.best_score_, 4))

rf_model_tuned = grid_search.best_estimator_
rf_model_pred = rf_model_tuned.predict(x_test)

from sklearn.metrics import accuracy_score
print("\nTuned Random Forest Accuracy:", accuracy_score(y_test, rf_model_pred))
```

```python
# Generating classification report for tuned Random Forest model
from sklearn.metrics import classification_report

print(classification_report(y_test, rf_model_pred))
```

```python
# Creating confusion matrix for tuned Random Forest Model
from sklearn.metrics import confusion_matrix

rf_cm_tuned = confusion_matrix(y_test, rf_model_pred)

sns.heatmap(rf_cm_tuned, annot=True, fmt='d', cmap='viridis')
plt.title('Tuned Random Forest Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
# calculating ROC-AUC score using prediction Probabilities
from sklearn.metrics import roc_auc_score

rf_prob=rf_model_tuned.predict_proba(x_test)[:,1]

roc_auc=roc_auc_score(y_test, rf_prob)

print("ROC-AUC Score:", roc_auc)
```

```python
# Plotting ROC Curve to visualize model performance
from sklearn.metrics import roc_curve

fpr, tpr, thresholds = roc_curve(y_test, rf_prob)

plt.figure(figsize=(7, 5))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC Curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=1.5, linestyle='--', label='Random Classifier')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate (Recall)')
plt.title('ROC Curve - Tuned Random Forest')
plt.legend(loc='lower right')
plt.grid(True, alpha=0.3)
plt.show()

# AUC closer to 1.0 = better model. Above 0.80 is considered good for churn prediction.
```

```python
# Cross-Validation — more reliable performance estimate than single train-test split
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(rf_model_tuned, x_train, y_train, cv=5, scoring='recall')

print("5-Fold Cross-Validation Recall Scores:", cv_scores.round(4))
print("Mean Recall:", round(cv_scores.mean(), 4))
print("Std Dev:", round(cv_scores.std(), 4))

# Low std dev means the model performs consistently across different data splits
```


### **Conclusion**
*   **Data Summary:** Analyzed 7,043 customers with a ~26.5% churn rate. Key preprocessing included converting `TotalCharges` to numeric and applying One-Hot Encoding to categorical features.
*   **Model Performance:**
    *   **Logistic Regression:** 74.7% accuracy, strong recall (82%) for churners.
    *   **Tuned Random Forest:** Optimized for recall (85%), achieving a high **ROC-AUC of 0.86**, indicating excellent class separation.
*   **Top Churn Drivers:** Tenure, Total Charges, Monthly Charges, Contract Type (Month-to-Month), and Internet Service (Fiber Optic).
